```python
sentence: str = "This is a python course"
use_vowels = []

words : list[str] = sentence.split(" ")

for word in words: 
    vowels = ["a", "e", "i", "o", "u", "ÿ"] 
    for letter in word: 
        if letter in vowels: 
            use_vowels.append(letter)
print(use_vowels)
```

```python
# what are functions 
# function take an input (argumnet), processs it, and return
#an output (return)

def shorten_names(names_list: list[str], chars: int = 3) -> list[str]: 
    """
    Returns a list of stringsthat are the first three characters of the word with a preiod appended
    Arges:
        names_list (list[str]): a list of strings that will be shortened
        chars (int): the number of characters to hsorten each string to. must be greater then 0. Defaults to 3"""
    if type(chars) != int: 
        raise TypeError("Chars must be an int")
    if chars < 0:
        raise ValueError("Chars must be greater than 0")
    names_list = [f"{name[:3]}." for name in names_list]
    return names_list
```

```python
bar_names: list[str] = ["Ottoman Empire", "France", "Spain"]

shorten_names(bar_names, chars = 2)
```
